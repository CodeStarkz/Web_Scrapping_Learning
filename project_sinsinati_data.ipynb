```
from bs4 import BeautifulSoup as bs
import requests as rqst
import pandas as pd
```

```
open_url="https://www.bbc.com/sport/tennis/scores-and-schedule"
```

```
raw_data_from_server= rqst.get(
    url=open_url
)
```

```
soup=bs(
    raw_data_from_server.content,
    "html.parser"
)
```

```
print(soup.prettify())
```
