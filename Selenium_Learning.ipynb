```
# Scrolling to a specific element
driver=webdriver.Chrome()
driver.maximize_window()
# visiting google com
driver.get("https://jobs.zs.com/all/jobs")
time.sleep(10)
#locating location bar
xpath_D_bar= '//*[@id="mat-select-4"]/div/div[2]/div'
time.sleep(3)
Department_bar=driver.find_element(By.XPATH,
                                 value=xpath_Department_bar)
element=driver.find_element(By.XPATH,'//*[@id="mat-expansion-panel-header-7"]/span[1]/mat-panel-title/p[1]')
print("Lets scroll to the element")
driver.execute_script("arguments[0].scrollIntoView();",element)
time.sleep(300)
```

```
# Vertical Scrolling(Use method scroll by)
"""
argument include two parameters
1. direction - Down(+ve) or Up(-ve)
2. amount - amount of pixels to scroll
"""
driver=webdriver.Chrome()
driver.maximize_window()
# visiting google com
driver.get("https://jobs.zs.com/all/jobs")
time.sleep(10)
#locating location bar
xpath_D_bar= '//*[@id="mat-select-4"]/div/div[2]/div'
time.sleep(3)

job_xpath=driver.find_element(By.XPATH,'//*[@id="link-job-8"]/span')
job_xpath.click()

# Scrolling down
time.sleep(5)
driver.execute_script("window.scrollBy(0,1200)")
time.sleep(30)
driver.quit()

```

```
# Horizontal Scrolling(Use method scroll by)
"""
argument include two parameters
1. direction - Down(+ve) or Up(-ve)
2. amount - amount of pixels to scroll
"""
driver=webdriver.Chrome()
driver.maximize_window()
# visiting google com
driver.get("https://jobs.zs.com/all/jobs")
time.sleep(10)
#locating location bar
xpath_D_bar= '//*[@id="mat-select-4"]/div/div[2]/div'
time.sleep(3)

job_xpath=driver.find_element(By.XPATH,'//*[@id="link-job-8"]/span')
job_xpath.click()

# Scrolling down
time.sleep(5)
driver.execute_script("window.scrollBy(500,1500)")
time.sleep(30)
driver.quit()
```

# Frames and IFrames
## It help us to embeded the one HTML item inside another.
### Frame is the older way, iFrame is the newer way.
###
AN iframe is an HTML element that embeds another HTML document within the current page. So changes made to the embedded document are reflected in the current page and vice versa. And each embedded document has its own history and can be navigated independently and DOM structure too.
