# The Script

```
driver=webdriver.Chrome()
driver.maximize_window()
driver.get("https://quotes.toscrape.com")
print(driver.title)
print(driver.current_url)
url_content_holder=rqst.get(driver.current_url)
soup=BeautifulSoup(
    url_content_holder.content,
    "html.parser"
)
quote_list=[]
author_list=[]
for i in (soup.find("div",class_="container").find_all("div",class_="quote")):
    #print(i.find("span",class_="text").text)
    quote_list.append(i.find("span",class_="text").text)
for i in (soup.find("div",class_="container").find_all("div",class_="quote")):
    author_list.append(i.find("small",class_="author").text)

df=pd.DataFrame({
    "Quotes":quote_list,
    "Author":author_list
})
driver.close()
df.head()
```

# Xpath
## Querry language that is used to navigate and locate element through xml and html

```
driver=webdriver.Chrome()
driver.maximize_window()
driver.get("https://www.google.com")
time.sleep(5)
#Locating the search bar
search_bar_xpath= '//*[@id="APjFqb"]'
search_bar=driver.find_element(By.XPATH,
                               value=search_bar_xpath)
# enter input
search_bar.send_keys("machine Learning")# to enter input
time.sleep(5)
# to clear the field
search_bar.clear()
time.sleep(5)
search_bar.send_keys("Advance machine learning")
time.sleep(20)
search_bar.send_keys(Keys.ENTER)

"""elements_xpath='//*[@id="_00ODaoLDAqWchvcP1rnrgQY_53"]'
elements=driver.find_element(By.XPATH,value=elements_xpath)
elements.click()"""
driver.quit()
```
