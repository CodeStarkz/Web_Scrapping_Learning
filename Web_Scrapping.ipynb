# What is WebScrapping
## Automated way to gathering data from websites, like a bot navigating through websites and collting the data on the basis of Predefined Rules

# What is WebCrawlling
## Web Crawlling is also know as Spidering, its the process systematic navigation over internet to discover index web pages.
## Web Crawller starts its journey with given url, and then it follows the links on the page to find more pages, in a recursive manner.


# Process Overview
1. GET request to targeted URL
2. Parse the HTML
3. Filter out the required data

# Client Server Architecture
## Client(Requestor) - Mobile, WebBrowser, Laptops
## Server(Responder) - WebServer(A dedicated system/Application to serve the requested data)

### HTTP Request - A message sent by Client to Server, to acess Resources. A Request consist of {"Request Line ": ["Method", Address, Version], Headers:"Used to convey additional information about the request and body", Body:"Data to be sent with the request"}

# HTTP Methods
## GET - Used to retrieve data from a server
## POST - Used to send data to a server
## PUT - Used to updated the existing resources (replacing the existing dat with new data)
## PATCH - Partial Update
## DELETE - Used to delete the resources from the server

# HTTP Status
## 1XX - Informational
## 2XX - Success
## 3XX - Redirection
## 4XX - Client Error
## 5XX - Server Error
