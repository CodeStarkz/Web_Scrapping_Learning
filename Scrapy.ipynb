## Defining Spider
### A program that navigate from one pages to anoyther pages to scrape data
## Spider navigates through web pages(a python class)
## Scrappers - Tool or function that extract data from a web pages

# Spider consist of
## - inherit Scrapy.spider as base class
## - unique name of spider
## - one or more base urls
## - And a parser
## - A name attribute that can be used to refer and run the spider

# Anatomy of Spiders
## Name Attribute (must be unique and it is used to run/execute the Spider)
## Start urls - entry point to start crawls
## Scrapy will automatically make a request to each urls in the list and call the parse() method over thr response received
## Parse() method- Is the default callback method to handle response
### It extracts data from the pages using CSS selector and XPATHs
### It can also yield new requests to other pages
## Response.follow()
### used to crawl internal links from one page to the next
### it is preffred over scrapy.Request() for internal navigation

# Types of Spider
## 1 Basic Spider
### This can be inherited from scrapy.spider
### Most commonly used
### Allow user to manually control  [Starting URLs, How pages are Parsed, How links are followed]
### Good for custom logic and flexibility
## 2 CrawlSpider (Rule based Spider)
### Inherit from scrapy.spiders.CrawlSpider
### Use rules to automate link-following behavior
## 3 XML feed Spider
### Used to parse XML feeds
### Handles large feeds efficiently without loading the whole document into memory
## 4 CSVFeedSpider
### Used to scrape CSV files,
### automatically splits rows and maps them to columns
# How to Decide, spider to use
### Basic Spider- in case of custom logic for crawling and parsing
### CrawlSpider - for follow links using patterns (pagination and category pages)
### XMLFeeedSpider - You are scrapping structured XML data
### CSVFedSpider - For scrapping CSV data source

```
# XML FeedSpider
from scrapy.spiders import XMLFeedSpider
class Myxmlfeedspider(XMLFeedSpider):
    name = 'myxmlfeed'
    allowed_domains = ['mysite.com']
    start_urls = ['http://www.mysite.com/']

    def parse_node(self, response, node) -> Any:
        yield {
            'title':node.find('title').text,
            'link' :node.find('link').text,
        }
```

```
# CSV feed spider
from scrapy.spiders import CSVFeedSpider
class Mycsvfeedspider(CSVFeedSpider):
    name = 'mycsvfeed'
    allowed_domains = ['mysite.com']
    start_urls = ['http://www.mysite.com/']
    def parse_row(self, response: Response, row: dict[str, str]) -> Any:
        yield {
            'name':row['name'],
            'price':row['price']
        }
```
