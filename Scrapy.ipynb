# Scrapy
## Scrapy is an open source web scraping and web crawling written in python. Designed for large scale scraping and data extraction from pages and asynchronous
## Comes with Built in Project Structure, Commandline tools, pipelines,etc.
## Ascyncronous - Extremely fast and efficient in comparision of scynchronous scrappers like BeautifulSoup and Requests
## Make it easy to crawl multiple pages and navigate through websites
## Buit-in data export : Export to JSON, CSV, XML or directly to databases
## Reusibility : User can organize and resue spiders and pipelines eaisly
## Provides User-Agent Rotation and Proxy Support to help avoid getting blocked

# To start project
### scrapy startproject
# Purpose of the structure
## 1 Enhances project scability
## 2 Inproves maintainability (modular in nature)
## 3 Component are reusable


### Adding a new website to scrape - Just add new file in spiders/
### Changing texport format - Tweak settings.py
### Validating Data - Use pipelines.py(Validation and Preprocessing)
### Dealing with CATCHAs and Blocking - Use middlewares.py
### Testing a single spider - Run only that spider with scrapy crawl
### Reusing data structure Define once in items.py and use across spiders

#### Run spider with command  -.. scrapy crawl <spider_name> -o output.json

## Defining Spider
### A program that navigate from one pages to anoyther pages to scrape data
## Spider navigates through web pages(a python class)
## Scrappers - Tool or function that extract data from a web pages
